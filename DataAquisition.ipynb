# Task 6

```python
gdp.rename(columns={'Country':'country'}, inplace=True)
gdp.head()
```

```python
literacy_rate.rename(columns={'name':'country'},inplace=True)
literacy_rate.head()
```

# Task 7

```python
Population["population"] = pd.to_numeric(Population["population"])

gdp["GDP"] = pd.to_numeric(gdp["GDP"])

internet_users["internet_users"] = pd.to_numeric(internet_users["internet_users"])

literacy_rate["literacy_rate"] = pd.to_numeric(literacy_rate["literacy_rate"])
```

```python
Population.dtypes
```

All datas are in integer types

# Task 8

```python
df = Population.merge(gdp, on='country', how='inner')
df = df.merge(internet_users, on='country', how='inner')
df = df.merge(literacy_rate, on='country', how='inner')

df.head()
```

# Task 9

```python
df["Internet Penetration (%)"] = round((df["internet_users"] / df["population"]) * 100, 2)
df["GDP Per Capita"] = round(df["GDP"] / df["population"], 2)
df.head()
```

# Task 10

```python
import sqlite3
```

```python
try:
  conn=sqlite3.connect("country_statistics.db")
  cursor=conn.cursor()
  print("Database created successfully")
except sqlite3.Error as error:
  print("Error connecting to database,e")
```

# Task 11

```python
df.to_sql(
    'country_statistics',
    conn,
    if_exists='replace',
    index=False
)
```

```python

pd.read_sql("SELECT * FROM country_statistics LIMIT 5", conn)
```

# Part F- Data Analysis Using SQL

```python
#11,
query = """
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
#12,
query = """
SELECT AVG(literacy_rate) AS average_literacy_rate
FROM country_statistics;
"""

pd.read_sql_query(query, conn)
```

```python
#13,
query = """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90;
"""

pd.read_sql_query(query, conn)
```

```python
#14,
query = """
SELECT country, "Internet Penetration (%)"
FROM country_statistics
ORDER BY "Internet Penetration (%)" DESC
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
#15,
query = """
SELECT SUM(population) AS total_population
FROM country_statistics;
"""

pd.read_sql_query(query, conn)
```

```python
#16,
query = """
SELECT country, "GDP Per Capita"
FROM country_statistics
ORDER BY "GDP Per Capita" DESC
LIMIT 1;
"""
pd.read_sql_query(query,conn)
```

# ETL Process Summary

Extracted country-related data from four different sources: CSV,
Excel, JSON, and XML files.

Cleaned the data by checking for missing values, duplicate records, standardizing column names, and converting fields to appropriate numeric data types.

Combined all datasets into a single table and created two additional metrics: Internet Penetration (%) and GDP Per Capita for better analysis.

Loaded the processed dataset into a SQLite database under the table country_statistics for efficient storage and querying.

The analysis highlighted differences in GDP, population, literacy, and internet usage across countries, providing a clear view of their overall development.
