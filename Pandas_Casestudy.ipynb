**5) Rename columns ("Title" > "Movie_Title")**


```python
df=df.rename(columns={"Title":"Movie_Title"})
df.columns

```

**6)Get unique genres**

```python
df["Genre"].str.split(",").explode().unique()
```

```python
#Number of Unique Genres
df["Genre"].str.split(",").explode().nunique()
```

**7) Extract number of genres per movie**

```python
df["Genre_Count"] = df["Genre"].str.count(",") + 1
df[["Movie_Title","Genre_Count"]]
```

**8) Handle missing values in 'Revenue (Millions)' and 'Metascore**


```python
#Finding missing values in Revenue and Metascore
df[["Revenue (Millions)","Metascore"]].isnull().sum()
```

```python
#Filling or handling null values
df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(df["Revenue (Millions)"].mean())

df["Metascore"] = df["Metascore"].fillna(df["Metascore"].mean())
df
```

```python
#Finally there is no null values
df[["Revenue (Millions)", "Metascore"]].isnull().sum()
```

**9) Top 5 movies by Rating**


```python
df.sort_values(by="Rating", ascending=False).head(5)[["Movie_Title", "Rating"]]
```

**10) Movies with highest/lowest revenue**

```python
#Movies with Highest revenue
df.nlargest(5, "Revenue (Millions)")[["Movie_Title","Revenue (Millions)"]]
```

```python
#Movies with lowest revenue
df.nsmallest(5, "Revenue (Millions)")[["Movie_Title","Revenue (Millions)"]]
```

**16) Create a new column Profitability:**

     a) Revenue divided by Duration

     b) Find top 10 most profitable movies

```python
#a,
df["Profitability"] = df["Revenue (Millions)"] / df["Runtime (Minutes)"]
df
```

```python
#b,
df.nlargest(10, "Profitability")[["Movie_Title","Profitability"]]
```

**17.Create a column that categorizes movies by rating:**

       a) Rating < 6: Poor

       b) 6 ≤ Rating < 7.5: Average

       c) Rating ≥ 7.5: Excellent

```python
df["Category"] = np.where(df["Rating"] < 6, "Poor",
                 np.where(df["Rating"] < 7.5, "Average", "Excellent"))
```

```python
df[["Movie_Title","Rating","Category"]]
```

**18)Create a custom function to score movies based on:**

        i) Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score

           Apply a function to dataset to assign the score

```python
def score(row):
    return row["Rating"] * 10 + row["Revenue (Millions)"]

df["Score"] = df.apply(score, axis=1)
df[["Movie_Title","Score"]]
```
