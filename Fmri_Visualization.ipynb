# The fmri dataset


```
It is a built-in dataset available in Seaborn.
It contains simulated functional Magnetic Resonance Imaging (fMRI) data
It used for visualization and analysis

```



````python
import matplotlib.pyplot as plt
````

````python
import seaborn as sns

fmri = sns.load_dataset("fmri")
````

````python
fmri.head()
````

````python
fmri[].value_counts()
````

````python
fmri.info()
````

````python
fmri.shape
````

# Lineplot

#Insight:

```

> Brain signal changes over time.
> Both stim and cue show similar signal patterns
```




````python
sns.lineplot(data=fmri, x="timepoint", y="signal", hue="event")
plt.show()
````

# Lineplot by region

````python
sns.lineplot(data=fmri, x="timepoint", y="signal", hue="region") #Brain signal varies over time in both regions.
                                                                 #The frontal and parietal regions show similar trends.
plt.show()
````

# Scatterplot
##insight:

```
The frontal and parietal regions show similar signal patterns across timepoints, with no clear separation between them.
```



````python
sns.scatterplot(data=fmri, x="timepoint", y="signal", hue="region")
plt.show()
````

# Boxplot
##Insight:


```
Box Plot (x="event", y="signal")

Both stim and cue have similar median signal values, with no major difference in their distributions.
```



````python
sns.boxplot(data=fmri, x="event", y="signal")
plt.show()
````

````python
sns.boxplot(data=fmri, x="region",y="signal") #The frontal and parietal regions have similar median signal values and distributions.
plt.show()
````

# Violin plot
##insight:

```
The frontal and parietal regions have similar signal distributions, with most values concentrated around the center
```



````python
sns.violinplot(data=fmri, x="region", y="signal")
plt.show()
````

# Histogram
##Insight:


```
Most brain signal values are concentrated near 0, with fewer observations at higher signal values
```

````python
sns.histplot(data=fmri, x="signal", kde=True)
plt.show()
````

# FacetGrid
##Insight:



# Relplot
##Insight:


```
The stim event produces a much stronger signal response than the cue event, peaking around timepoint 5–6 before gradually decreasing
```



````python
sns.relplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event",
    kind="line"
)
````

# Catplot
##Insight:


```
The stim event has greater signal variability and more high-value outliers than the cue event
```



````python
sns.catplot(
    data=fmri,
    x="event",
    y="signal",
    kind="box"
)
````

# Displot
##Insight:


```
Most observations cluster near 0, while a few high signal values create a long right tail.
```



````python
sns.displot(
    data=fmri,
    x="signal",
    kde=True
)
````

# Joint plot
##Insight:


```
There is a clear time-dependent pattern in the signal, with a peak around timepoints 5–7 followed by a gradual decrease
```



````python
sns.jointplot(
    data=fmri,
    x="timepoint",
    y="signal",
    kind="scatter"
)
````

# Lineplot with style
##Insight:


```
The stim event produces a stronger signal response than the cue event in both brain regions, while the parietal region consistently shows higher signal values than the frontal region
```



````python
sns.lineplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event",
    style="region"
)
plt.show()
````

# Scatterplot with size
##Insight:


```
The scatter plot shows that the stim event produces a significantly stronger brain signal than the cue event.
The signal reaches its maximum around timepoints 5–7 and gradually decreases afterward.
The cue event remains close to the baseline with minimal variation. While there is some variability across subjects,
 the overall response pattern is consistent.
```




````python
sns.scatterplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event",
    size="subject"
)
plt.show()
````

````python
corr = fmri.corr(numeric_only=True)
corr
````

# Heatmap
##Insight:


```
This heatmap shows the correlation between the two numerical variables: timepoint and signal.

The signal has a weak negative correlation with timepoint (r = -0.28),

indicating that signal values tend to decrease slightly as time progresses.

```



````python
sns.heatmap(fmri.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
````
