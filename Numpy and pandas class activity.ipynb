```python
#3                                             #np.where() is used for conditional selection or replacement.
import numpy as np
a= np.random.randint(10,100,15)
print("Original:", a)
count = np.sum(a> 50)

a = np.where(a > 50, -1, a)

print("Modified:", a)
print("Count:", count)
```
