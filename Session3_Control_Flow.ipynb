# **Loop control statements**
consist of

*   Break - it will exit the entire loop,Stops the loop completel
*   Continue -Skips the current iteration and moves to the next one
*   Pass   -pass does nothing,It is a placeholder,means sometimes if we need a block,because we forget the code or anything,so that time we can use pass,if we didn't use pass at that time,it makes error in python





```python
for i in range(5): #here it break the loop when i==3,so output is 3
  if i ==3:
     break
  print(i)
```

```python
for i in range(5): #continue is used for skip one iteration,here when i==3,what will skip and print other all
  if i ==3:
    continue
  print(i)
```

```python
for i in range(5): #just pass,nothing happens
  if i ==3:
   pass
  print(i)
```
