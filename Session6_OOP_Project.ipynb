```python
class Phone:
    def __init__(self, brand, model, price):
        self.brand = brand      # Instance Variable
        self.model = model      # Instance Variable
        self.price = price      # Instance Variable

    def display(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Price:", self.price)


# Object Creation
phone1 = Phone("Samsung", "Galaxy S24", 75000)
phone2 = Phone("Apple", "iPhone 16", 90000)

# Calling Method
phone1.display()

print()

phone2.display()
```

```python
class Camera(Phone):
    def __init__(self, brand, model, price, camera_mp):
        super().__init__(brand, model, price)
        self.camera_mp = camera_mp

    def show_camera(self):
        print("Camera:", self.camera_mp, "MP")


camera1 = Camera("Samsung", "Galaxy S24", 75000, 108)


camera1.display()
camera1.show_camera()

```

```python
class SmartPhone(Camera):
    def __init__(self, brand, model, price, camera_mp, storage):
        super().__init__(brand, model, price, camera_mp)
        self.storage = storage

    def show_storage(self):
        print("Storage:", self.storage, "GB")



phone1 = SmartPhone("Samsung", "Galaxy S24", 75000, 108, 256)


phone1.display()
phone1.show_camera()
phone1.show_storage()
```
