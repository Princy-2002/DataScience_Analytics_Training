Sale Price is the Target column,and it has 4 null values,

```python
df = df.dropna(subset=["Sale Price"]).copy()
```

```python
median_cols = [
    "No of Bathrooms",
    "Flat Area (in Sqft)",
    "Lot Area (in Sqft)",
    "Area of the House from Basement (in Sqft)",
    "Latitude",
    "Longitude",
    "Living Area after Renovation (in Sqft)"
]

for col in median_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df["Zipcode"] = df["Zipcode"].fillna(df["Zipcode"].mode()[0])
```

```python
columns = [
    "Sale Price",
    "No of Bathrooms",
    "Flat Area (in Sqft)",
    "Lot Area (in Sqft)",
    "Area of the House from Basement (in Sqft)",
    "Latitude",
    "Longitude",
    "Living Area after Renovation (in Sqft)"
]
```

```python
for i in columns:
  plt.figure(figsize=(5,5))
  plt.boxplot(df[i])
  plt.title(i)
  plt.show()
```

```python
from scipy.stats.mstats import winsorize

winsor_cols = [
    "No of Bathrooms",
    "Flat Area (in Sqft)",
    "Lot Area (in Sqft)",
    "Area of the House from Basement (in Sqft)",
    "Living Area after Renovation (in Sqft)"
]

for col in winsor_cols:
    df[col] = winsorize(df[col], limits=[0.02, 0.02])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in winsor_cols:
    plt.figure(figsize=(4,4))
    sns.boxplot(y=df[col])
    plt.title(col)
    plt.show()
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[["Bad", "Fair", "Okay", "Good", "Excellent"]]
)

df_new[["Condition of the House"]] = encoder.fit_transform(
    df_new[["Condition of the House"]]
)

# Convert float to integer
df_new["Condition of the House"] = df_new["Condition of the House"].astype(int)
```
