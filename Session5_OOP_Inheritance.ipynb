```python
class BankAccount:                       #__init__() is a special method (constructor) that is automatically called when an object is created.
  accStatus= "Active"                    #It is used to initialize the object's attributes.
  def __init__(self,owner,accno,balance):         #2f is used inside an f-string to format a number to 2 decimal places
   self.owner=owner
   self.accno =accno
   self.balance = balance

  def deposit(self,amount):
    if amount >0:
      self.balance +=amount
      print(f"Deposited {amount:.2f}. Balance:{self.balance:.2f}")
    else:
      print("Not possible ")

  def withdraw(self,amount):
    if amount >0:
       if self.balance >= amount:
        self.balance -=amount
        print(f"withdraw {amount:.2f}. Balance {self.balance:.2f}")
       else:
        print("Insufficient balance")
    else:
      print("Can't process")

  def display(self):
     print(f" Owner:{self.owner} Accno:{self.accno} Balance:{self.balance:.2f}")

cust1= BankAccount("Alan",145570,2000)
cust2= BankAccount("Lena",145523,2500)
cust3= BankAccount("Rejin",145590,3800)

cust1.display()
cust1.deposit(500)
cust1.display()
cust2.display()

cust1.withdraw(5000)
cust1.deposit(-2)
cust1.deposit(1000)
cust1.withdraw(-100)



```

```python
class SavingsAccount(BankAccount):
  def __init__(self,owner,accno,balance,interest_rate,min_balance):
    super().__init__(owner,accno,balance)
    self.interest_rate = interest_rate
    self.min_balance = min_balance

  def calculate_interest(self):
    interest =self.balance * self.interest_rate
    self.balance +=interest

  def withdraw(self,amount):
     if self.balance - amount >= self.min_balance:
       super().withdraw(amount)
     else:
        print("Insufficient balance")

cust3 =SavingsAccount("Rejin", 145590,3800,0.05,1000)
cust3.deposit(500)
cust3.calculate_interest()
cust3.display()
```

```python
class PremiumSavingsAccount(SavingsAccount):

    def __init__(self, owner, accno, balance,
                 interest_rate, min_balance, bonus_rate):

        super().__init__(
            owner,
            accno,
            balance,
            interest_rate,
            min_balance
        )

        self.bonus_rate = bonus_rate

    def calculate_bonus(self):
        bonus = self.balance * self.bonus_rate
        self.balance += bonus
        print(f"Bonus added: {bonus:.2f}")

acc1 = PremiumSavingsAccount("Anu", 12345, 10000, 0.05, 1000, 0.02)
acc1.display()
acc1.calculate_interest()
acc1.calculate_bonus()
acc1.withdraw(3000)
acc1.display()
```
