```python
class BankAccount:
  accStatus= "Active"
  def __init__(self,owner,accno,balance):
   self.owner=owner
   self.accno =accno
   self.balance = balance

  def deposit(self,amount):
    if amount >0:
      self.balance +=amount
      print(f"Deposited {amount:.2f}. Balance:{self.balance:.2f}")
    else:
      print("Not possible ")

  def withdraw(self,amount):
    if amount >0:
       if self.balance >= amount:
        self.balance -=amount
        print(f"withdraw {amount:.2f}. Balance {self.balance:.2f}")
       else:
        print("Insufficient balance")
    else:
      print("Can't process")


  def balance_enquiry(self):
        print(f"Current Balance: {self.balance:.2f}")

  def display(self):
     print(f" Owner:{self.owner} Accno:{self.accno} Balance:{self.balance:.2f}")

cust1= BankAccount("Meenu",145570,2000)
cust2= BankAccount("Lekshmi",145523,2500)
cust3= BankAccount("Anjana",145590,3800)

cust1.display()
cust1.deposit(600)
cust1.display()
cust2.display()
cust2.balance_enquiry()

cust1.withdraw(5000)
cust1.deposit(-2)
cust1.deposit(1000)
cust1.withdraw(-100)
cust1.balance_enquiry()
```

```python
region = "Mavelikara"
humidity = 72
temperature = 28


def calculate_avg_temp(temp):
    return temp

def calculate_avg_humidity(hum):
    return hum


if humidity > 70:
    condition = "Rainy"
elif temperature > 30:
    condition = "Sunny"
else:
    condition = "Cloudy"


print("Region:", region)
print("Temperature:", temperature, "°C")
print("Humidity:", humidity, "%")
print("Weather Condition:", condition)

avg_temp = calculate_avg_temp(temperature)
avg_humidity = calculate_avg_humidity(humidity)

print("Average Temperature:", avg_temp, "°C")
print("Average Humidity:", avg_humidity, "%")
```
