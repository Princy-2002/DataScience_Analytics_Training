# Module 5-Business Functions & Reporting

````python
#1,get_product_price()
def get_product_price(product_id, products_list):

    for product in products_list:
        if product["Product_ID"] == product_id:
            return product["Price"]

    return None
````

````python
print(get_product_price("P08", products))
````

````python
#2,get_customer_name()
def get_customer_name(customer_id, customers_list):

    for customer in customers_list:
        if customer["Customer_ID"] == customer_id:
            return customer["Name"]

    return None
````

````python
print(get_customer_name("C01", customers))
````

````python
#3,calculate_order_value()
def calculate_order_value(price, quantity):
    return price * quantity
````

````python
print(calculate_order_value(4999, 1))
````

````python

def classify_order(value, high_limit=3000, low_limit=1000):

    if value > high_limit:
        return "High"

    elif value >= low_limit:
        return "Medium"

    else:
        return "Low"
````

````python
print(classify_order(4999))
print(classify_order(2000))
print(classify_order(500))
````

````python
print(classify_order(2500, high_limit=2000, low_limit=500))
````

````python
#5,generate_invoice()
def generate_invoice(order, products_list, customers_list):

    customer_name = get_customer_name(
        order["Customer_ID"],
        customers_list
    )

    price = get_product_price(
        order["Product_ID"],
        products_list
    )

    order_value = calculate_order_value(
        price,
        order["Quantity"]
    )

    invoice = f"""
----- UrbanCart Invoice -----
Order ID      : {order["Order_ID"]}
Customer      : {customer_name}
Product ID    : {order["Product_ID"]}
Quantity      : {order["Quantity"]}
Price         : ₹{price}
Order Value   : ₹{order_value}
Payment Mode  : {order["Payment_Mode"]}
Status        : {order["Status"]}
-----------------------------
"""

    return invoice
````

````python
print(generate_invoice(orders[0], products, customers))
````

````python
#6,calculate_total_revenue()-for delivered orders only
def calculate_delivered_revenue(orders_list, products_list):

    total = 0

    for order in orders_list:

        if order["Status"] != "Delivered":
            continue

        price = get_product_price(
            order["Product_ID"],
            products_list
        )

        total += calculate_order_value(
            price,
            order["Quantity"]
        )

    return total
````

````python
print("Delivered Revenue:",
      calculate_delivered_revenue(orders, products))
````

````python
#7,top customer by spending
def top_customer_by_spend(orders_list, products_list, customers_list):

    customer_spending = {}

    for order in orders_list:

        if order["Status"] != "Delivered":
            continue

        price = get_product_price(
            order["Product_ID"],
            products_list
        )

        order_value = calculate_order_value(
            price,
            order["Quantity"]
        )

        customer_id = order["Customer_ID"]

        if customer_id not in customer_spending:
            customer_spending[customer_id] = 0

        customer_spending[customer_id] += order_value

    top_customer_id = None
    highest_spending = 0

    for customer_id, spending in customer_spending.items():

        if spending > highest_spending:
            highest_spending = spending
            top_customer_id = customer_id

    return get_customer_name(
        top_customer_id,
        customers_list
    )
````

````python
print("Top Customer:",
      top_customer_by_spend(orders, products, customers))
````

````python
#8,Apply Multiple Discounts
def apply_offer(price, *discounts):

    final_price = price

    for discount in discounts:
        final_price = final_price - (final_price * discount / 100)

    return final_price
````

````python
print(apply_offer(1000, 10, 5))
````

````python
#3,Total Customers + Premium/Regular Split
total_customers = len(customers)

premium_count = 0
regular_count = 0

for customer in customers:
    if customer["Membership"] == "Premium":
        premium_count += 1
    else:
        regular_count += 1

print("Total Customers:", total_customers)
print("Premium Customers:", premium_count)
print("Regular Customers:", regular_count)
````

````python
#4,Total Orders + Order Status Split
total_orders = len(orders)

delivered_count = 0
pending_count = 0
cancelled_count = 0

for order in orders:
    if order["Status"] == "Delivered":
        delivered_count += 1
    elif order["Status"] == "Pending":
        pending_count += 1
    elif order["Status"] == "Cancelled":
        cancelled_count += 1

print("Total Orders:", total_orders)
print("Delivered Orders:", delivered_count)
print("Pending Orders:", pending_count)
print("Cancelled Orders:", cancelled_count)
````

````python
#5,Total Revenue
total_revenue = 0

for order in orders:
    if order["Status"] != "Delivered":
        continue

    for product in products:
        if order["Product_ID"] == product["Product_ID"]:
            order_value = product["Price"] * order["Quantity"]
            total_revenue += order_value

print("Total Revenue:", total_revenue)
````

````python
#6,Average Order Value
average_order_value = total_revenue / delivered_count

print("Average Order Value:", round(average_order_value, 2))
````

````python
#7,Top Customer
customer_spending = {}

for order in orders:
    if order["Status"] != "Delivered":
        continue

    for product in products:
        if order["Product_ID"] == product["Product_ID"]:
            order_value = product["Price"] * order["Quantity"]

            customer_id = order["Customer_ID"]

            if customer_id not in customer_spending:
                customer_spending[customer_id] = 0

            customer_spending[customer_id] += order_value

print("Customer Spending:", customer_spending)
````

````python
highest_spending = 0
top_customer_id = None

for customer_id, spending in customer_spending.items():
    if spending > highest_spending:
        highest_spending = spending
        top_customer_id = customer_id

print("Top Customer ID:", top_customer_id)
print("Highest Spending:", highest_spending)
````

````python
top_customer_name = ""

for customer in customers:
    if customer["Customer_ID"] == top_customer_id:
        top_customer_name = customer["Name"]

print("Top Customer:", top_customer_name)
````

````python
#8,No of out of stock products
out_of_stock_count = 0

for product in products:
    if product["Stock"] == 0:
        out_of_stock_count += 1

print("Out-of-Stock Products:", out_of_stock_count)
````

````python
#9,count quantity by category
category_quantity = {}

for order in orders:
    if order["Status"] != "Delivered":
        continue

    for product in products:
        if order["Product_ID"] == product["Product_ID"]:

            category = product["Category"]

            if category not in category_quantity:
                category_quantity[category] = 0

            category_quantity[category] += order["Quantity"]

print("Category-wise Delivered Quantity:")
print(category_quantity)
````

````python
#Highest Category
most_popular_category = None
highest_quantity = 0

for category, quantity in category_quantity.items():
    if quantity > highest_quantity:
        highest_quantity = quantity
        most_popular_category = category

print("Most Popular Category:", most_popular_category)
print("Delivered Quantity:", highest_quantity)
````
